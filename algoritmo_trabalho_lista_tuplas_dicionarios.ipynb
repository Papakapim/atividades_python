```python
# quarta questão
numeros = []
pares = []
impares = []
for i in range(10):
  par_ou_impar = int(input('Número: '))
  numeros.append(par_ou_impar)
  if par_ou_impar%2 == 0:
    pares.append(par_ou_impar)
  else:
    impares.append(par_ou_impar)
print('Números digitados:', end = ' ')
for n in numeros:
  print(f'{n}', end = ' ')
print('\nNúmeros pares:', end = ' ')
for p in pares:
  print(f'{p}', end = ' ')
print('\nNúmeros impares:', end = ' ')
for i in impares:
  print(f'{i}', end = ' ')
```

```python
# quinta questão
numeros = []
maior, menor = float('-inf'), float('inf')
for i in range(8):
  num = int(input('Digite um número: '))
  numeros.append(num)
  if maior < numeros[i]:
    maior  = numeros[i]
    posicao = numeros.index(maior)
  elif menor > numeros[i]:
    menor = numeros[i]
    posicao2 = numeros.index(menor)
print(f'Maior: {maior}\nPosição do maior: {posicao+1}º lugar')
print(f'Menor: {menor}\nPosição do menor: {posicao2+1}º lugar9')
```

```python
# sexta questão
nomes = []
print('Digite "FIM" para parar de escrever nomes')
while True:
  nome = input('Digite um nome: ')
  if nome in nomes:
    print('Você não pode adicionar nomes já presentes na lista de nomes')
    continue
  elif nome == 'FIM':
    print('Programa encerrado.')
    break
  nomes.append(nome)
print('Nomes:=', end = ' ')
for i in nomes:
  print(f'{i}', end = ' ')
```

```python
# oitava questão
tupla = ()
quantidade5, pares = 0, 0
for i in range(6):
  try:
    num = int(input('Digite um número: '))
    tupla = tupla + (num,)
  except ValueError:
    print('Escreva apenas números inteiros')
for i, numero in enumerate(tupla):
  if numero == 5:
    quantidade5 = quantidade5 + 1
    if quantidade5 == 1:
      posicao = i
  elif numero % 2 == 0:
    pares = pares + 1
if quantidade5 > 0:
  print(f'A primeira aparição do número 5 foi em {posicao+1}º lugar\nApareceu {quantidade5} vezes')
print(f'Foram escritos {pares} números pares')
```

```python
# nona questão
nome = input('Qual seu nome?\n')
turma = input('Qual sua turma?\n')
nota1 = float(input('Qual foi sua primeira nota? '))
nota2 = float(input('Qual foi sua segunda nota? '))
while 10 < nota1 or 10 < nota2 or 0 > nota1 or 0 > nota2:
  print('Digite notas de 0 a 10')
  nota1 = float(input('Qual foi sua primeira nota? '))
  nota2 = float(input('Qual foi sua segunda nota? '))
dados = {'Nome': nome,'Turma': turma, 'Nota 1': nota1, 'nota2': nota2}
if 10 >= (nota1 + nota2)/2 >= 7:
  situacao = 'Aprovado'
elif 7 > (nota1+nota2)/2 >= 5:
  situacao = 'Recuperação'
elif 5 > (nota1+nota2)/2 >=0:
  situacao = 'Reprovado'
print(f'Nome: {dados['Nome']}')
print(f'Turma: {dados['Turma']}')
print(f'Primeira nota: {dados['Nota 1']}')
print(f'Segunda nota: {dados['nota2']}')
print(f'Média: {(dados['Nota 1']+dados['nota2'])/2}')
print(f'Situação: {situacao}')
```

```python
# décima primeira questão
palavras = {}
for i in range(8):
  palavra = input('Digite uma palavra: ')
  if palavra in palavras: palavras[palavra] = palavras[palavra] + 1
  else: palavras[palavra] = 1
print('Palavras digitadas:')
for p in palavras: print(f'{p} ({palavras[p]} vezes)')
```

```python
# décima segunda questão
produtos = {'shampoo':100, 'condicionador':100,'sabonete':75,'hidratante':50}
rodar = True
while rodar:
  produto = input('Produto: ').lower()
  while True:
    try:
      quantidadevendida = int(input('Quantidade vendida: '))
      break
    except ValueError:
      print('Digite números')
  if produto in produtos:
    if quantidadevendida <= produtos[produto] and quantidadevendida > 0:
      produtos[produto] -= quantidadevendida
      rodar2 = True
      while rodar2:
        continuar = input('Digite "s" para sim e "n" para não\nDeseja continuar?\n').lower()
        if continuar == 'n':
          rodar2 = False
          rodar = False
        elif continuar == 's':
          rodar2 = False
        else:
          print('Digite "s" ou "n"')
    else:
      print('Quantidade excedente ou quantidade inválida')
  else:
    print('Produto não encontrado')
print('Estoque:')
for i in produtos:
  print(f'Produto: {i}\nQuantidade: {produtos[i]}')


```

```python
# décima terceira questão
lista_alunos = []
for i in range(5):
  aluno = {}
  nome = input('Nome do aluno: ')
  aluno['nome'] = nome
  while True:
    try:
      nota1 = float(input('Primeira nota: '))
      if nota1 < 0 or nota1 > 10:
        raise ValueError
      aluno['primeira nota'] = nota1
      break
    except ValueError:
      print('Valor inválido digitado')
  while True:
    try:
      nota2 = float(input('Segunda nota: '))
      if nota2 < 0 or nota2 > 10:
        raise ValueError
      aluno['segunda nota'] = nota2
      break
    except ValueError:
      print('Valor inválido digitado')
  media = (nota1+nota2)/2
  aluno['média'] = media
  lista_alunos.append(aluno)
print('===========INFORMAÇÕES=========== \n')
for aluno in lista_alunos:
  for chave, valor in aluno.items():
    print(f'{chave}: {valor}')

```

```python
# décima quarta questão
lista = []
for i in range(5):
  tupla_produtos = ()
  codigo = input('Digite o código do produto:\n')
  tupla_produtos += (codigo,)
  nome = input('Digite o nome do produto:\n')
  tupla_produtos += (nome,)
  while True:
    try:
      preco = float(input('Digite o preço do produto:\n'))
      if preco <= 0:
        raise ValueError
      break
    except ValueError:
        print('Digite números positivos')
  tupla_produtos += (preco,)
  lista.append(tupla_produtos)
print('\n\n')
print('===========PARTE DO CLIENTE==========\n\n')
principal = True
contagem = 0
while principal:
  rodaresse = True
  for produto in lista:
        while rodaresse:
          leitura = input('Digite o código do produto que deseja acessar\n')
          if leitura == produto[0]:
            print(f'Produto: {produto[1]}\nPreço: {produto[2]}')
            rodaresse = False
            contagem += 1
  if contagem == 0:
    print('Código inválido')
  rodar = True
  while rodar:
    print('Digite "s" para sim e "n" para não')
    opcao = input('Deseja ver outro produto?\n')
    if opcao == 's':
      rodar = False
    elif opcao == 'n':
      principal = False
    else:
      continue
print('Código encerrado hehehe wow br br patapim')
```

```python
# décima quinta questão
lista_itens = []
maiores = []
maior = float('-inf')
quantidade_itens = 0
for i in range(5):
  item = {}
  codigo = input(f'Código do produto {i+1}: ')
  item['código'] = codigo
  nome = input(f'Nome do produto {i+1}: ')
  item['nome'] = nome
  while True:
    try:
      quantidade = int(input(f'Quantidade do produto {i+1}: '))
      if quantidade < 0:
        print('Digite valores inteiros positivos')
        continue
      item['quantidade'] = quantidade
      quantidade_itens += quantidade
      break
    except ValueError:
      print('Digite números')
  while True:
    try:
      preco = float(input(f'Preço do produto {i+1}: '))
      if preco <= 0:
        print('Digite um preço válido')
        continue
      item['preço'] = preco
      break
    except ValueError:
      print('Digite um valor numérico')
  estoque = preco * quantidade
  item['valor do estoque'] = estoque
  if maior < estoque:
    maior = item['valor do estoque']
  maiores.append(estoque)
  lista_itens.append(item)
print('\n\n\n=======ITENS========')
for item in lista_itens:
  for chave, valor in item.items():
    print(f'{chave}: {valor}')
soma = 0
for s in maiores:
  soma += s
print('INFORMAÇÕES ADICIONAIS: ')
tupla = (quantidade_itens,maior,soma)
print(f'Quantidade total de itens: {tupla[0]}')
print(f'Estoque total de itens: {tupla[2]}')
print(f'Maior valor de estoque por item: {tupla[1]}')




```
