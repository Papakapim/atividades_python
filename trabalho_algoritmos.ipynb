```python
# questão 5
nota1 = float(input("Qual foi a sua primeira nota? "))
nota2 = float(input("Qual foi a sua segunda nota? "))
media = (nota1+nota2)/2
if 10 >= media >= 7:
  print("Aprovado!")
elif 7 > media >= 5:
  print("Recuperação")
elif 5 > media >= 0:
  print("Reprovado.")
else:
  print("Nota inválida digitada")
print(f"Média: {media}")
```

```python
# questão 14
print("----- tente acertar o número secreto! -----\n")
numero_secreto = 7
loop = 0
while True:
  tentativa = int(input("Digite seu palpite: "))
  loop += 1
  if tentativa > numero_secreto:
    print("Errou!, tente digitar um número menor")
  elif tentativa < numero_secreto:
    print("Errou! tente digitar um número maior")
  else:
    print(f"Parabéns!!! Você acertou em {loop} tentativas")
    break
```
