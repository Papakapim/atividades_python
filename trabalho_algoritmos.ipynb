```python
# quarta questão
horas = int(input('Digite a quantidade de horas estacionadas: '))
if 0 <= horas <= 2:
  print('Total a pagar: 8 reais')
elif 2 < horas <= 5:
  print(f'Total a pagar: {8 + 3*(horas - 2)} reais')
elif 5 < horas:
  print(f'Total a pagar: {17 + 2*(horas - 5)} reais')
```

```python
# quinta questão
print('Preço do ingresso: 25 reais')
pagamento = int(input('Pagamento do ingresso: '))
soma_p = pagamento
while pagamento < 0:
  print('Você digitou um valor inválido\nTente novamente')
  pagamento = int(input('Pagamento do ingresso: '))
  soma_p = pagamento
while soma_p < 25:
  print('Pagamento insufiente\nDigite um valor a ser adicionado ao pagamento anterior')
  pagamento = int(input('Pagamento do ingresso: '))
  soma_p += pagamento
print(f'Total: {soma_p} reais\nTroco: {soma_p - 25} reais')
```

```python
# questão 9

saldo = 500
for i in range(6):
  pergunta = int(input("Escolha:\n1 - depósito \n2 - saque\nDigite um desses números aqui: "))
  while pergunta != 1 and pergunta != 2:
     print("Dígito inválido.")
     pergunta = int(input("Escolha:\n1 - depósito\n2 - saque\nDigite um desses números aqui: "))
  if pergunta == 1:
    deposito = int(input('O quanto você quer depositar? '))
    saldo += deposito
    print(f'Seu novo saldo é de {saldo}')
  elif pergunta == 2:
    saque = int(input('O quanto você quer sacar? '))
    if saque > saldo:
       print('Saldo insuficiente para saque.')
    else:
      saldo -= saque
      print(f'Saque concluído com sucesso!\nSaldo atual: {saldo}')
```

```python
# questão 11
numero = int(input('Digite um número: '))
divisores = 0
for i in range(numero):
  if numero%(i+1) == 0:
    divisores += 1
if divisores == 2:
  print(f'{numero} é primo!')
else: print(f'Primo: Não\nDivisores: {divisores}')
```

```python
# questão 13
A = int(input('Digite um número: '))
B = int(input('Digite outro número: '))
divisores_comuns = 0
while A < 0:
     A = int(input(f'Digite um número positivo no lugar do {A}: '))
     while A >= B:
      B = int(input(f'{A} é maior ou igual que {B}\nDigite um número maior que {A}: '))
for i in range(A):
  if B%(i+1) == A%(i+1) == 0:
    divisores_comuns += 1
if divisores_comuns == 1:
  print(f'{A} e {B} são primos entre si')
else:
    print(f'{A} e {B} não são primos entre si, e possuem {divisores_comuns} divisores em comum')
```

```python
# questão 15
saldo, entradas, retiradas, retiradas_r, total_entrou, total_saiu = 200, 0, 0, 0, 0, 0
opcao = input("1 - Depositar\n2 - Sacar/Retirar\n0 - encerrar/Saida\nOpção escolhida: ")
while opcao != '1' and opcao != '2' and opcao != '0':
  opcao = input("Você digitou valores inválidos\n1 - Depositar\n2 - Sacar/Retirar\nOpção: ")
while opcao != '0':
  while opcao == '1':
     print(f'Saldo atual: {saldo}')
     deposito = int(input("Quanto deseja depositar? "))
     if deposito >= 0:
       entradas += 1
       saldo += deposito
       total_entrou += deposito
       print(f'Saldo atual: {saldo}')
       break
     else:
       print("Valor inválido digitado")
  opcao = input("1 - Depositar\n2 - Sacar/Retirar\n0 - encerrar/Saida\nOpção escolhida: ")
  while opcao == '2':
     print(f"Saldo atual: {saldo}")
     saque = int(input('Quanto deseja sacar? '))
     if saque <= 0:
        print("Valor inválido digitado")
     else:
       if saque <= saldo:
         saldo -= saque
         retiradas += 1
         total_saiu += saque
         print(f'Novo saldo: {saldo}')
         break
       else:
         retiradas_r += 1
         print('Saldo insuficiente')
print('Programa finalizado!')
print(f"Saldo final: {saldo}")
print(f'Quantidade de entradas/depósitos: {entradas}')
print(f"Depósito total: {total_entrou}")
print(f'Quantidade de retiradas/saques: {retiradas}')
print(f"Quantidade de retiradas/saques recusadas(os): {retiradas_r}")
print(f"Saques finais: {total_saiu}")
```
